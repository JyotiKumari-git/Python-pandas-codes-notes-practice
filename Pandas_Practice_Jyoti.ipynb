```python
import pandas as pd

Data1={
    'Name':['Mohan','Devika','Isha','Yash'],
    'Age':[22,28,31,19],
    'City':['Delhi','Pune','Agra','UP'],
    'Salary':[20000,15000,35000,18000]
}
df=pd.DataFrame(Data1)
print(df)
```

```python
#9.aggregation and Groupby-->
#want to know the count of Team :
df['Name'].value_counts()
```

```python
#groupby--
df.groupby('City')['Salary'].sum()
```

```python
#groupby+agg
#if we want the count of name and mean & avg of salry we write-->
df.groupby('City').agg({'Salary':'mean','Name':'count'})
```

```python
# 10.concatenate and merge the dataframe-->(JOINS)
data_A=pd.DataFrame({'ID':[1,2,3],'Name':['A','B','C']})
print(data_A)
```

```python
data_B=pd.DataFrame({'ID':[1,2,3,4],'Score':[88,96,77,79]})
print(data_B)
```

```python
pd.concat([data_A,data_B],axis=0) #row wise concat
```

```python
pd.concat([data_A,data_B],axis=1)   #column wise concat
```

```python
#merge=join
pd.merge(data_A,data_B,how='inner' ,on='ID') #put coomon cloumn ID
#by default inner
```

```python
pd.merge(data_A,data_B,how='inner' ,left_on='ID'right_on='ID',)
#for multiple common column
```
